```python
import pandas as pd
import numpy as np
from pathlib import Path

pd.set_option("display.max_columns", None)
```

```python
from pathlib import Path


# Load datasets
demographics = pd.read_csv(r"C:\Users\User\Downloads\demographics_clean.csv")
vitals = pd.read_csv(r"C:\Users\User\OneDrive\Documents\vitals_clean.csv")
adverse_events = pd.read_csv(r"C:\Users\User\OneDrive\Documents\adverse_event_clean.csv")
lab_results = pd.read_csv(r"C:\Users\User\OneDrive\Documents\lab_results_clean.csv")

print("All datasets loaded successfully.")
```

```python
# Check the first 5 rows of each dataset

print("DEMOGRAPHICS")
display(demographics.head())

print("\nVITAL SIGNS")
display(vitals.head())

print("\nADVERSE EVENTS")
display(adverse_events.head())

print("\nLAB RESULTS")
display(lab_results.head())
```

```python
# Check rows, columns, data types and missing values

datasets = {
    "Demographics": demographics,
    "Vitals": vitals,
    "Adverse Events": adverse_events,
    "Lab Results": lab_results
}

for name, df in datasets.items():
    print("\n" + "=" * 60)
    print(name.upper())
    print("=" * 60)

    print("\nShape:", df.shape)

    print("\nColumn Names:")
    print(df.columns.tolist())

    print("\nData Types:")
    print(df.dtypes)

    print("\nMissing Values:")
    print(df.isnull().sum())
```

```python
# Standardize column names

for name, df in datasets.items():
    
    df.columns = (
        df.columns
        .str.strip()
        .str.upper()
        .str.replace(" ", "_")
    )

print("Column names standardized successfully.")
```

```python
# Remove unnecessary spaces from text columns

for name, df in datasets.items():
    
    text_columns = df.select_dtypes(include="object").columns
    
    for column in text_columns:
        df[column] = df[column].str.strip()

print("Text values cleaned successfully.")
```

```python
# Check duplicate records

for name, df in datasets.items():
    
    duplicate_count = df.duplicated().sum()
    
    print(f"{name}: {duplicate_count} duplicate record(s)")
```

```python
# Add duplicate flags

demographics["DUPLICATE_FLAG"] = demographics.duplicated(
    subset=["SUBJECT_ID"],
    keep=False
)

vitals["DUPLICATE_FLAG"] = vitals.duplicated(
    keep=False
)

adverse_events["DUPLICATE_FLAG"] = adverse_events.duplicated(
    keep=False
)

lab_results["DUPLICATE_FLAG"] = lab_results.duplicated(
    keep=False
)

print("Duplicate flags created.")
```

```python
# Display duplicate records if they exist

print("DEMOGRAPHICS DUPLICATES")
display(
    demographics[
        demographics["DUPLICATE_FLAG"] == True
    ]
)

print("\nVITALS DUPLICATES")
display(
    vitals[
        vitals["DUPLICATE_FLAG"] == True
    ]
)

print("\nADVERSE EVENTS DUPLICATES")
display(
    adverse_events[
        adverse_events["DUPLICATE_FLAG"] == True
    ]
)

print("\nLAB RESULTS DUPLICATES")
display(
    lab_results[
        lab_results["DUPLICATE_FLAG"] == True
    ]
)
```

```python
# Convert demographic age to numeric

demographics["AGE"] = pd.to_numeric(
    demographics["AGE"],
    errors="coerce"
)


# Convert vital sign columns to numeric

numeric_vital_columns = [
    "SYSTOLIC_BP",
    "DIASTOLIC_BP",
    "PULSE",
    "TEMP_C"
]

for column in numeric_vital_columns:
    
    numeric_vital_columns = [
        "SYSTOLIC_BP",
        "DIASTOLIC_BP",
        "HEART_RATE",
        "TEMP_C"
    ]

    for column in numeric_vital_columns:
        vitals[column] = pd.to_numeric(
            vitals[column],
            errors="coerce"
        )


# Convert lab result to numeric

lab_results["RESULT_NUMERIC"] = pd.to_numeric(
    lab_results["RESULT"],
    errors="coerce"
)

print("Numeric conversion completed.")
```

```python
# Convert date columns to datetime

demographics["ENROLL_DATE"] = pd.to_datetime(
    demographics["ENROLL_DATE"],
    errors="coerce"
)

vitals["VISIT_DATE"] = pd.to_datetime(
    vitals["VISIT_DATE"],
    errors="coerce"
)

adverse_events["START_DATE"] = pd.to_datetime(
    adverse_events["START_DATE"],
    errors="coerce"
)

adverse_events["END_DATE"] = pd.to_datetime(
    adverse_events["END_DATE"],
    errors="coerce"
)

lab_results["VISIT_DATE"] = pd.to_datetime(
    lab_results["VISIT_DATE"],
    errors="coerce"
)

print("Date conversion completed.")
```

```python
# Missing value summary after cleaning

for name, df in datasets.items():
    
    missing_values = df.isnull().sum()
    
    missing_values = missing_values[
        missing_values > 0
    ]
    
    print("\n" + "=" * 50)
    print(name.upper())
    print("=" * 50)
    
    if len(missing_values) == 0:
        print("No missing values found.")
    else:
        print(missing_values)
```

```python
# Standardize text values

demographics["SEX"] = (
    demographics["SEX"]
    .str.upper()
    .replace({
        "MALE": "M",
        "FEMALE": "F"
    })
)

adverse_events["SEVERITY"] = (
    adverse_events["SEVERITY"]
    .str.upper()
)

print("Important text values standardized.")
```

```python
# Check for missing age

demographics["DQ_AGE_MISSING"] = demographics["AGE"].isna()


# Check for age outside the expected range

demographics["DQ_AGE_RANGE"] = (
    demographics["AGE"].notna()
    &
    ~demographics["AGE"].between(18, 100)
)


print("Age validation completed.")
```

```python
# Display age-related data quality issues

age_issues = demographics[
    (demographics["DQ_AGE_MISSING"] == True)
    |
    (demographics["DQ_AGE_RANGE"] == True)
]

display(age_issues)
```

```python
# Check for invalid sex values

demographics["DQ_SEX_INVALID"] = (
    demographics["SEX"].notna()
    &
    ~demographics["SEX"].isin(["M", "F"])
)


print("Sex validation completed.")
```

```python
# Flag missing enrollment dates

demographics["DQ_ENROLL_DATE_MISSING"] = (
    demographics["ENROLL_DATE"].isna()
)

print("Enrollment date validation completed.")
```

```python
# Flag missing vital sign values

vitals["DQ_SYSTOLIC_MISSING"] = (
    vitals["SYSTOLIC_BP"].isna()
)

vitals["DQ_DIASTOLIC_MISSING"] = (
    vitals["DIASTOLIC_BP"].isna()
)

vitals["DQ_HEART_RATE_MISSING"] = (
    vitals["HEART_RATE"].isna()
)

vitals["DQ_TEMP_MISSING"] = (
    vitals["TEMP_C"].isna()
)


print("Missing vital sign checks completed.")
```

```python
# Check systolic blood pressure range

vitals["DQ_SYSTOLIC_RANGE"] = (
    vitals["SYSTOLIC_BP"].notna()
    &
    ~vitals["SYSTOLIC_BP"].between(70, 250)
)

print("Systolic BP validation completed.")
```

```python
# Check diastolic blood pressure range

vitals["DQ_DIASTOLIC_RANGE"] = (
    vitals["DIASTOLIC_BP"].notna()
    &
    ~vitals["DIASTOLIC_BP"].between(40, 150)
)

print("Diastolic BP validation completed.")
```

```python
# Check pulse range

vitals["DQ_PULSE_RANGE"] = (
    vitals["HEART_RATE"].notna()
    &
    ~vitals["HEART_RATE"].between(30, 220)
)

print("Pulse validation completed.")
```

```python
# Check temperature range

vitals["DQ_TEMP_RANGE"] = (
    vitals["TEMP_C"].notna()
    &
    ~vitals["TEMP_C"].between(34, 42)
)

print("Temperature validation completed.")
```

```python
# Check blood pressure logic

vitals["DQ_BP_LOGIC"] = (
    vitals["SYSTOLIC_BP"].notna()
    &
    vitals["DIASTOLIC_BP"].notna()
    &
    (
        vitals["DIASTOLIC_BP"]
        >=
        vitals["SYSTOLIC_BP"]
    )
)

print("Blood pressure logic check completed.")
```

```python
# Create a combined filter for vital sign issues

vital_issues = vitals[
    (
        vitals["DQ_SYSTOLIC_MISSING"]
        |
        vitals["DQ_DIASTOLIC_MISSING"]
        |
        vitals["DQ_HEART_RATE_MISSING"]
        |
        vitals["DQ_TEMP_MISSING"]
        |
        vitals["DQ_SYSTOLIC_RANGE"]
        |
        vitals["DQ_DIASTOLIC_RANGE"]
        |
        vitals["DQ_PULSE_RANGE"]
        |
        vitals["DQ_TEMP_RANGE"]
        |
        vitals["DQ_BP_LOGIC"]
    )
]

display(vital_issues)
```

```python
# Add enrollment date to vital signs data

vitals = vitals.merge(
    demographics[
        ["SUBJECT_ID", "ENROLL_DATE"]
    ],
    on="SUBJECT_ID",
    how="left"
)

print("Enrollment date added to vital signs data.")

display(vitals.head())
```

```python
# Check for missing visit dates

vitals["DQ_VISIT_DATE_MISSING"] = (
    vitals["VISIT_DATE"].isna()
)


# Check if visit happened before enrollment

vitals["DQ_VISIT_BEFORE_ENROLL"] = (
    vitals["VISIT_DATE"].notna()
    &
    vitals["ENROLL_DATE_y"].notna()
    &
    (
        vitals["VISIT_DATE"]
        <
        vitals["ENROLL_DATE_y"]
    )
)

print("Visit date validation completed.")
```

```python
# Display records with visit date issues

visit_date_issues = vitals[
    vitals["DQ_VISIT_DATE_MISSING"]
    |
    vitals["DQ_VISIT_BEFORE_ENROLL"]
]

display(
    visit_date_issues[
        [
            "SUBJECT_ID",
            "VISIT_DATE",
            "ENROLL_DATE_y",
            "DQ_VISIT_DATE_MISSING",
            "DQ_VISIT_BEFORE_ENROLL"
        ]
    ]
)
```

```python
# Flag missing adverse event terms

adverse_events["DQ_AE_TERM_MISSING"] = (
    adverse_events["AE_TERM"].isna()
)

print("AE term validation completed.")
```

```python
# Flag missing adverse event start dates

adverse_events["DQ_START_DATE_MISSING"] = (
    adverse_events["START_DATE"].isna()
)

print("AE start date validation completed.")
```

```python
# Flag missing severity

adverse_events["DQ_SEVERITY_MISSING"] = (
    adverse_events["SEVERITY"].isna()
)

print("AE severity missing check completed.")
```

```python
# Flag invalid severity values

adverse_events["DQ_SEVERITY_INVALID"] = (
    adverse_events["SEVERITY"].notna()
    &
    ~adverse_events["SEVERITY"].isin(
        ["MILD", "MODERATE", "SEVERE"]
    )
)

print("AE severity validation completed.")
```

```python
# Check adverse event date logic

adverse_events["DQ_DATE_LOGIC"] = (
    adverse_events["START_DATE"].notna()
    &
    adverse_events["END_DATE"].notna()
    &
    (
        adverse_events["END_DATE"]
        <
        adverse_events["START_DATE"]
    )
)

print("AE date logic validation completed.")
```

```python
# Display adverse event records with data quality issues

ae_issues = adverse_events[
    adverse_events["DQ_AE_TERM_MISSING"]
    |
    adverse_events["DQ_START_DATE_MISSING"]
    |
    adverse_events["DQ_SEVERITY_MISSING"]
    |
    adverse_events["DQ_SEVERITY_INVALID"]
    |
    adverse_events["DQ_DATE_LOGIC"]
]

display(ae_issues)
```

```python
# Flag missing laboratory results

lab_results["DQ_RESULT_MISSING"] = (
    lab_results["RESULT"].isna()
)

print("Lab result missing check completed.")
```

```python
# Flag invalid non-numeric laboratory results

lab_results["DQ_RESULT_INVALID"] = (
    lab_results["RESULT"].notna()
    &
    lab_results["RESULT_NUMERIC"].isna()
)

print("Lab result validation completed.")
```

```python
# Flag missing laboratory units

lab_results["DQ_UNIT_MISSING"] = (
    lab_results["UNIT"].isna()
)

print("Lab unit validation completed.")
```

```python
# Create default clinical flag

lab_results["CLINICAL_FLAG"] = "NORMAL"


# Identify records with valid result and reference range

valid_lab_range = (
    lab_results["RESULT_NUMERIC"].notna()
    &
    lab_results["NORMAL_LOW"].notna()
    &
    lab_results["NORMAL_HIGH"].notna()
)


# Flag low results

lab_results.loc[
    valid_lab_range
    &
    (
        lab_results["RESULT_NUMERIC"]
        <
        lab_results["NORMAL_LOW"]
    ),
    "CLINICAL_FLAG"
] = "LOW"


# Flag high results

lab_results.loc[
    valid_lab_range
    &
    (
        lab_results["RESULT_NUMERIC"]
        >
        lab_results["NORMAL_HIGH"]
    ),
    "CLINICAL_FLAG"
] = "HIGH"


print("Clinical lab flags created.")
```

```python
# Show data quality issues

lab_data_issues = lab_results[
    lab_results["DQ_RESULT_MISSING"]
    |
    lab_results["DQ_RESULT_INVALID"]
    |
    lab_results["DQ_UNIT_MISSING"]
]

print("LAB DATA QUALITY ISSUES")
display(lab_data_issues)


# Show clinical abnormalities separately

lab_clinical_review = lab_results[
    lab_results["CLINICAL_FLAG"].isin(
        ["LOW", "HIGH"]
    )
]

print("CLINICALLY ABNORMAL RESULTS")
display(lab_clinical_review)
```

```python
# Create an empty list to store all data queries

queries = []
```

```python
def add_query(subject_id, dataset, field, issue):

    queries.append({
        "SUBJECT_ID": subject_id,
        "DATASET": dataset,
        "FIELD": field,
        "ISSUE": issue,
        "STATUS": "Open"
    })
```

```python
for _, row in demographics.iterrows():

    if row["DUPLICATE_FLAG"]:
        add_query(
            row["SUBJECT_ID"],
            "Demographics",
            "SUBJECT_ID",
            "Duplicate subject record"
        )

    if row["DQ_AGE_MISSING"]:
        add_query(
            row["SUBJECT_ID"],
            "Demographics",
            "AGE",
            "Age is missing"
        )

    if row["DQ_AGE_RANGE"]:
        add_query(
            row["SUBJECT_ID"],
            "Demographics",
            "AGE",
            "Age is outside the expected range"
        )
```

```python
for _, row in vitals.iterrows():

    if row["DUPLICATE_FLAG"]:
        add_query(
            row["SUBJECT_ID"],
            "Vitals",
            "Record",
            "Duplicate vital sign record"
        )

    if row["DQ_SYSTOLIC_MISSING"]:
        add_query(
            row["SUBJECT_ID"],
            "Vitals",
            "SYSTOLIC_BP",
            "Systolic BP is missing or invalid"
        )

    if row["DQ_PULSE_RANGE"]:
        add_query(
            row["SUBJECT_ID"],
            "Vitals",
            "PULSE",
            "Pulse is outside the expected range"
        )

    if row["DQ_BP_LOGIC"]:
        add_query(
            row["SUBJECT_ID"],
            "Vitals",
            "BLOOD_PRESSURE",
            "Diastolic BP is greater than or equal to systolic BP"
        )

    if row["DQ_VISIT_BEFORE_ENROLL"]:
        add_query(
            row["SUBJECT_ID"],
            "Vitals",
            "VISIT_DATE",
            "Visit date occurs before enrollment date"
        )
```

```python
for _, row in adverse_events.iterrows():

    if row["DQ_AE_TERM_MISSING"]:
        add_query(
            row["SUBJECT_ID"],
            "Adverse Events",
            "AE_TERM",
            "Adverse event term is missing"
        )

    if row["DQ_SEVERITY_INVALID"]:
        add_query(
            row["SUBJECT_ID"],
            "Adverse Events",
            "SEVERITY",
            "Invalid severity value"
        )

    if row["DQ_DATE_LOGIC"]:
        add_query(
            row["SUBJECT_ID"],
            "Adverse Events",
            "END_DATE",
            "End date occurs before start date"
        )
```

```python
for _, row in lab_results.iterrows():

    if row["DQ_RESULT_MISSING"]:
        add_query(
            row["SUBJECT_ID"],
            "Lab Results",
            "RESULT",
            "Laboratory result is missing"
        )

    if row["DQ_RESULT_INVALID"]:
        add_query(
            row["SUBJECT_ID"],
            "Lab Results",
            "RESULT",
            "Laboratory result is invalid"
        )

    if row["DQ_UNIT_MISSING"]:
        add_query(
            row["SUBJECT_ID"],
            "Lab Results",
            "UNIT",
            "Laboratory unit is missing"
        )
```

```python
query_listing = pd.DataFrame(queries)

# Create query IDs
query_listing.insert(
    0,
    "QUERY_ID",
    [f"Q{i:03d}" for i in range(1, len(query_listing) + 1)]
)

# Display final query listing
query_listing
```
